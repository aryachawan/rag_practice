```python
import os
from dotenv import load_dotenv
load_dotenv()

from langchain_core.documents import Document

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
# initializing llm
from langchain_groq import ChatGroq
llm = ChatGroq(
    model_name = "llama-3.1-8b-instant"
)
llm
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
```

Importing main vectorstore library

```python
from langchain_core.vectorstores import InMemoryVectorStore
vectorstore = InMemoryVectorStore(
    embedding=embedding
)
```

```python
# sample docs (irl would be extracted from pdfs,docs,txtfiles,dbs etc)
documents = [
    Document(
        page_content="I had chocolate chips and a vanilla milkshake while watching a movie last night. The weather was pleasant and the streets were quiet.",
        metadata={"source": "tweet"}
    ),
    Document(
        page_content="Researchers announced a new battery technology that may increase electric vehicle range significantly. Early tests have shown promising results.",
        metadata={"source": "news"}
    ),
    Document(
        page_content="The website tutorial explained how to build a simple calculator using HTML, CSS, and JavaScript. It included code examples and exercises.",
        metadata={"source": "website"}
    ),
    Document(
        page_content="A user asked how to improve laptop performance. Several replies suggested upgrading RAM and replacing the hard drive with an SSD.",
        metadata={"source": "forum"}
    ),
    Document(
        page_content="I tried a new coffee shop downtown today. The cappuccino was excellent and the atmosphere was calm and relaxing.",
        metadata={"source": "tweet"}
    ),
    Document(
        page_content="Stock markets rose after positive economic reports were released. Analysts expect volatility to remain high this quarter.",
        metadata={"source": "news"}
    ),
    Document(
        page_content="A guide described the basics of Python programming including variables, loops, functions, and object-oriented concepts.",
        metadata={"source": "website"}
    ),
    Document(
        page_content="Someone reported issues with a software update causing crashes. Community members shared temporary fixes and workarounds.",
        metadata={"source": "forum"}
    ),
    Document(
        page_content="I spent the afternoon reading a mystery novel while eating popcorn. The plot twists were completely unexpected.",
        metadata={"source": "tweet"}
    ),
    Document(
        page_content="Scientists discovered a new species of marine life during a deep-sea exploration mission. Further studies are planned.",
        metadata={"source": "news"}
    )
]
```

```python
# adding docs to vectorstore
vectorstore.add_documents(documents=documents)
```

```python
result = vectorstore.similarity_search("How is the weather forecast?",k=2)
print(result)
```
