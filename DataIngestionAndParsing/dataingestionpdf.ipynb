Handling PDF issues

```python
# Handling blank spaces example
raw_pdf_text = """Company Financial Report


     The finanical performance for fiscal year 2024
     shows significant growt in profitability.


     Revenue increased by 25%.

  The company's efficiency improved due to workflow optimization.

  Page 1 of 10
"""

# Space cleaning function
def clean_text(text):
    text = " ".join(text.split())
    return text

cleaned = clean_text(raw_pdf_text)
print("Before cleaning:")
print(repr(raw_pdf_text))
print("After cleaning:")
print(repr(cleaned))
```

Special class to handle PDF Processing

```python
from typing import List
from langchain_core.documents import Document
# Class to implement pdf processing
class SmartPDFProcessor:
    # Constructor 
    def __init__(self,chunk_size=1000,chunk_overlap=100):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap,
            separators=[" "]
        )
    
    def process_pdf(self,pdf_path:str) -> List[Document]:
        # Loading pdf
        loader = PyPDFLoader(pdf_path)
        pages = loader.load()

        # Processing each page
        processed_chunks=[]

        for page_num,page in enumerate(pages):
            # cleaning text
            cleaned_text = self._clean_text(page.page_content)

            # skipping empty page based on no of characters
            if len(cleaned_text.strip()) < 50:
                continue

            # creating chunks with more metadata
            chunks = self.text_splitter.create_documents(
                texts=[cleaned_text],
                metadatas=[{
                    **page.metadata,
                    "page": page_num+1,
                    "total_pages": len(pages),
                    "chunk_method": "smart_pdf_processor",
                    "char_count": len(cleaned_text)
                }]
            )
            processed_chunks.extend(chunks)
        
        return processed_chunks
    
    def _clean_text(self,text: str) -> str:
        # removing white spaces
        text = " ".join(text.split())
        return text
```

```python
# INSTANCE OF SMART PDF PROCESSOR
preprocessor = SmartPDFProcessor()
```

```python
# PROCESSING A PDF USING SMARTPDFPROCESSOR 
try:
    chunks=preprocessor.process_pdf("data/pdf_files/Scnotes.pdf")
    print(f"No of chunks created: {len(chunks)}")

    if chunks:
        print("Sample chunk metadata: ")
        print(chunks[0].page_content)
        for key,value in chunks[0].metadata.items():
            print(f"{key}: {value}")
except Exception as e:
    print(f"Error: {e}")
```
